#integrating the dictionary in loops and functions


```python
Student={
   "number1":{
    "name":"Akash",
    "marks":90
   },
   "number2":{
    "name":"Manish",
    "marks":95
   }
}
```

```python
print(Student["number1"]["marks"])
```

```python
Student["number3"]={"name":"azhakiri","marks":85}
```

```python
print(Student["number3"]["name"], Student["number3"]["marks"])
```

```python
Student["number2"]["marks"]=80
print(Student["number2"]["marks"])
```

```python
game=["volleyball","cricket","football"]
wanted_items=["volleyball","bat","football"]
list=dict(zip(game,wanted_items))
print(list)
```

```python
list1=[1,3,5,7,9]
dict1={i:i**2 for i in list1}
print(dict1)
```

```python
manavargal = {}

def add_student(name, marks):
    manavargal[name] = marks

add_student("vels", 90)

print(manavargal)
```

```python
players={
    "dhoni":"cricketer",
    "ronaldo":"footballer",
    "pv sindhu":"badminton player"
}
def view_players():
    for player, profession in players.items():
        print(f"{player} is a {profession}")

view_players()
```
