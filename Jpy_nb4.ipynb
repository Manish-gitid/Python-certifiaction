#using loops in dict

```python
students={
    "Manish":100,
    "Akash":70,
    "keerthi":90,
    "jebastin":80,
    "Subash":30
}
```

```python
for name in students.keys():
    print(name)
```

```python
for marks in students.values():
    print(marks)
```

```python
for name,marks in students.items():
    print(name,marks)
```

```python
for name,marks in students.items():
    if marks>=80:
        print(name,marks)
```

```python
count=0
for name,marks in students.items():
    if marks<40:
        count+=1
print("Number of students who failed:", count)
```

```python
total=0
for marks in students.values():
    total+=marks
print("Total marks:", total)
```

```python
print(max(students.items()))
```
